```python
#Download Data
import gdown
import os
import zipfile

file_id = "1GqQgpK5RDB3Nd72rHoucf0MQ1ECbdBzJ"

zip_path = "dataset.zip"
extract_dir = "dataset"

# Download file
print("Downloading...")
gdown.download(
    id=file_id,
    output=zip_path,
    quiet=False
)

# Unzip
print("Unzipping...")

os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(extract_dir)

print("Done!")

# Count WAV
wav_files = [
    os.path.join(root, file)
    for root, _, files in os.walk(extract_dir)
    for file in files
    if file.lower().endswith(".wav")
]

print(f"Số file .wav: {len(wav_files)}")
print("5 file đầu:")

for f in wav_files[:5]:
    print(f)
```
